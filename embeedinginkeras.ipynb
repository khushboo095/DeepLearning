```python
import numpy as np
word=['I am learning Deep Learning',



'Deep Learning is interesting',



'I love Machine Learning',



'Python is easy to learn',



'Neural networks are powerful',



'I am studying Data Science',



'TensorFlow is used in Deep Learning',



'NLP helps computers understand language',



'AI is changing the world',



'I want to become a Data Scientist']
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts(word)
```

```python
len(tokenizer.word_index)
```

```python
from keras.utils import pad_sequences
```

```python
sequences = tokenizer.texts_to_sequences(word)
sequences
```

```python
sequences = pad_sequences(sequences,padding='post')
```

```python
sequences
```

```python
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
# model = Sequential()
# model.add(Embedding(18,output_dim=2,input_length=5))

# model.summary()


from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding

model = Sequential([
    Embedding(input_dim=36, output_dim=2, input_length=5)
])

model.build(input_shape=(None, 7))
model.summary()
```

```python
model.compile( optimizer='adam', metrics=['accuracy'],  loss='binary_crossentropy')
```

```python
pred = model.predict(sequences)
print(pred)
```

```python
print("Sequences:", sequences)
print("Shape:", sequences.shape if hasattr(sequences, "shape") else "Not a NumPy array")

print("\nEmbedding configuration:")
for layer in model.layers:
    print(layer.name, layer.get_config())
```
